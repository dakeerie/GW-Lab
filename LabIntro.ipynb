# Analysing gravitational wave data
This project requires students to work in pairs or small groups to complete a data challenge to detect gravitational wave events and analyse them to discover their astrophysical parameters. This project will be carried out in the Python programming language using a series of Jupyter notebooks.

To date the LIGO and Virgo detectors have detected dozens of binary black hole mergers, a handful of neutron star-black hole mergers, and a couple of binary neutron star mergers. In this lab you will be provided with a segment of gravitational wave data. You must work together to identify gravitational wave signals in this data and perform parameter estimation on them to learn about the astrophysical sources which created them.


## Getting started
This project will be performed in Python, generally through the use of Jupyter Notebooks on the https://jupyter.physics.gla.ac.uk server. For this project you should use the IPython kernel "GravWaves". There are three ipython notebooks already present which will introduce you to several of the key concepts in gravitational wave data analysis. Remember to make a copy of these to your own namespace before you work through them so your results can be saved. Make sure to update the paths which point to the data files.

These include:
   * **signal_processing.ipynb**, which introduces fourier transforms, noise, windowing, computing power spectral densities, and generating waveforms.
   * **signal_detection.ipynb**, which introduces the concept of gravitational wave detection, computing SNRs, and matched filtering.
   * **bayesian_inference.ipynb**, which introduces the basics of parameter inference, including Bayes theorem and coding up a likelihood function.

Everyone should work their way through the signal processing notebook. **After this, you should decide whether you want to focus on detection or parameter estimation for the rest of your lab**. Ideally you should make sure there is at least one person doing detection, and one person doing parameter estimation in your group. The lab is designed so that both parts can be done in parallel. Some already-detected events are provided for the parameter estimation team to get started on. However, if your detection team is successful, they will identify a range of additional signals in their data which they can pass to the parameter estimation team in order to learn more about the astrophysical source of the signal.

For the remainder of the lab sessions you will work through the data challenge to detect or run parameter estimation on gravitational wave data. The lab is designed to be partially open ended, so it should not take you the full 9 weeks to complete. You will use the skills you have learned during the data challenge to come up with your own idea of a small project to continue at the end (such as incorporating data from more gravitational wave detectors, or inferring an extended set of parameters, or trying out the skills you have learned on real gravitational wave data... the choices are (almost) endless!). Make sure to talk your idea through with a demonstrator before you start it. Additional useful resources such as tutorials and relevant papers can be found in the following sections.

**You will not have time to do both the detection and parameter estimation parts of the data challenge by yourself, so decide who will focus on what by the end of lab 1**

### Useful papers
Some useful papers to read and reference are given below. Remember that you can look up the references given in these papers too! When you come to write your report you will want to dig a bit deeper in the area you are interested in.

Most LIGO-Virgo collaboration papers can also be easily found at http://papers.ligo.org.

#### Basic Introduction / Background reading
   * [A guide to LIGO-Virgo detector noise and extraction of transient gravitational-wave signals](https://arxiv.org/abs/1908.11170) by B. P. Abbott _et al_. A very nice overview of the basics (see also [this summary](https://cplberry.com/2020/02/09/gw-data-guides/#data-analysis-guide) of the content).
   * [Observation of Gravitational Waves from a Binary Black Hole Merger](http://adsabs.harvard.edu/abs/2016PhRvL.116f1102A) by B. P. Abbott _et al_, Phys. Rev. Lett., 116, 061102 (2016), [arXiv:1602.03837](https://arxiv.org/abs/1602.03837) (the paper describing the first gravitational wave detection GW150914)
   * "The Basic Physics of the binary Black Hole Merger GW150914" by LIGO Scientific Collaboration and Virgo Collaboration, https://onlinelibrary.wiley.com/doi/epdf/10.1002/andp.201600209
   * [GW170817: Observation of Gravitational Waves from a Binary Neutron Star Inspiral](http://adsabs.harvard.edu/abs/2017PhRvL.119p1101A) by B. P. Abbott _et al_, Phys. Rev. Lett., 119, 161101 (2017), arXiv:1710.05832 (a paper describing the first detection of gravitational waves from a binary neutron star coalescence).
   * [GWTC-1: A Gravitational-Wave Transient Catalog of Compact Binary Mergers Observed by LIGO and Virgo during the First and Second Observing Runs](http://adsabs.harvard.edu/abs/2018arXiv181112907T) by B. P. Abbott _et al,_ https://arxiv.org/abs/1811.12907 (the catalogue of gravitational wave signals detected during the LIGO/Virgo O1 and O2 observing runs).
